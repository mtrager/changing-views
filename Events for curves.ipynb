# Visual events for space curves

We consider a curve $C = \{F_1 = F_2 = 0\}$ in $\mathbb P^3$, the complete intersection of two surfaces. For degrees $(d_1, d_2)$ it has degree $n = d_1 d_2$ and genus $g = \frac12 d_1 d_2 (d_1 + d_2 - 4) + 1$.

```sage
from visual_events.surfaces import random_surface
from visual_events.space_curves import *
from visual_events.grassmannian import dimension_and_degree, order_and_class
```

```sage
K = GF(32003)
R.<x0,x1,x2,x3> = K[]
F1 = random_surface(R, 2)
F2 = random_surface(R, 3)
n = F1.degree() * F2.degree()
g = n * (F1.degree() + F2.degree() - 4) / 2 + 1
print("degree", n, "genus", g)
```

## Tangential surface

The union of the tangent lines of $C$; its degree is $2n + 2g - 2$.

```sage
print("(dim, deg):", dimension_and_degree(tangential_surface(F1, F2)))
```

## Trisecant and edge curves in ${\rm Gr}(1,\mathbb P^3)$

Trisecant lines meet $C$ in three points. By Cayley's formula, the ruled surface they sweep out has degree $\frac{(n-2)((n-1)(n-3) - 3g)}{3}$. Edge lines join two points of $C$ whose tangent lines meet. The edge computation is a slow elimination (more than 20 minutes here); Table 1 of [Kohn–Sturmfels–Trager](https://arxiv.org/abs/1707.01877) gives degree 54 for this curve.

```sage
print("trisecant (dim, deg):", dimension_and_degree(trisecant_curve(F1, F2)))
```

```sage
run_slow = False
if run_slow:
    print("edge (dim, deg):", dimension_and_degree(edge_curve(F1, F2)))
```

## Chow hypersurface and secant congruence

The lines meeting $C$ form a hypersurface section of ${\rm Gr}(1,\mathbb P^3)$ given by the Chow form, of degree $n$. Its singular locus is the secant congruence, of order $\binom{n-1}{2} - g$ and class $\binom{n}{2}$.

```sage
chow = chow_hypersurface(F1, F2)
dim, deg = dimension_and_degree(chow)
print("Chow hypersurface: dimension", dim, "| degree of Chow form", deg / 2)
print("secant (order, class):", order_and_class(secant_congruence(chow)))
```
