We consider a curve $C$ in $\mathbb P^3$ that is the intersection of two surfaces $F_1, F_2$.

```python
K = GF(9973)
R.<x0,x1,x2,x3> = K['x0,x1,x2,x3']

degree1 = 2
degree2 = 3

#F1 = general_surface(degree1)
#F2 = general_surface(degree2)

F1 = ideal(x0^3 + x1^3 + x2^3 + x3^3)
F2 = ideal(x0^2 + x1^2 + x2^2 + x3^2)

C = F1 + F2

print C
print C.hilbert_polynomial()
```

We compute the tangential surface $\mathcal T(C)$ in $\mathbb P^3$

```python
R1.<x0,x1,x2,x3,y0,y1,y2,y3> = K[]
X = matrix(R,[[x0],[x1],[x2],[x3]])

F1=F1.change_ring(R1)
F2=F2.change_ring(R1)
C=C.change_ring(R1)

Y = matrix(R1,[[y0],[y1],[y2],[y3]])

jmat = (jacobian(F1.gens()[0], (x0,x1, x2, x3))).stack(
    (jacobian(F2.gens()[0], (x0,x1, x2, x3))))

jmat_y = jmat.stack(matrix(R1,[[y0,y1,y2,y3]]))

#incidence = C + ideal(jmat_y.minors(3))
incidence = C + ideal((jmat*Y).list())
incidence = incidence.substitute(x0=1)

tangential = incidence.elimination_ideal((x1,x2,x3)).change_ring(K[y0,y1,y2,y3])
print "Dimension: ", tangential.dimension()
print "Degree: ", tangential.gens()[0].degree()
print tangential
```

Trisecant surface as a curve $\mathcal D^\ell(C)$ in ${\rm Gr}(1,\mathbb P^3)$

```python
R1.<x0, x1, x2, x3, p01, p02, p03, p12, p13, p23,t> = K[]

F1s = F1.change_ring(R1)
F2s = F2.change_ring(R1)
X = matrix(R1,[[x0],[x1],[x2],[x3]])
Q1 = matrix(R1,[[1,0,-p12,-p13]]).transpose()
Q2 = matrix(R1,[[0, 1, p02, p03]]).transpose()
Xline = Q1+t*Q2

dict_poly = dict(zip(X.list(),Xline.list()))

poly1 = (F1s.subs(dict_poly).gens()[0]).polynomial(t)
poly2 = (F2s.subs(dict_poly).gens()[0]).polynomial(t)


S0 = ideal(subrescoefs(poly1, poly2, 0)).change_ring(R1)
S1 = ideal(subrescoefs(poly1, poly2, 1)).change_ring(R1)
S2 = subrescoefs(poly1, poly2, 2)
leading_coeffs = ideal(poly1.coefficients()[poly1.degree()],
                       poly2.coefficients()[poly2.degree()]).change_ring(R1)


plucker = ideal(1*p23 - p02*p13 + p03*p12)
tritangent = plucker + S0 + S1 + S2
tritangent = tritangent.saturation(leading_coeffs)[0]
tritangent = tritangent.homogenize(p01)
tritangent = tritangent.change_ring(K[p01, p02, p03, p12, p13, p23])

print "Dimension: ", tritangent.dimension()-1
print "Degree: ", tritangent.hilbert_polynomial().coefficients()[1]
print tritangent
```

Edge surface as a curve $\mathcal E^\ell(C)$ in ${\rm Gr}(1,\mathbb P^3)$

```python
R1.<x0,x1,x2,x3,y0,y1,y2,y3,p01, p02, p03, p12, p13, p23> = K[]

X = matrix(R1,[[x0], [x1], [x2], [x3]])
Y = matrix(R1,[[y0], [y1], [y2], [y3]])

XY = X.augment(Y)
XYmin = XY.minors(2)

F1=F1.change_ring(R1)
F2=F2.change_ring(R1)
F1y=F1.subs(x0=y0,x1=y1,x2=y2,x3=y3)
F2y=F2.subs(x0=y0,x1=y1,x2=y2,x3=y3)
C=C.change_ring(R1)

jmatx = (jacobian(F1.gens()[0], (x0,x1, x2, x3))).stack(
    (jacobian(F2.gens()[0], (x0,x1, x2, x3))))
jmaty = (jacobian(F1y.gens()[0], (y0,y1, y2, y3))).stack(
    (jacobian(F2y.gens()[0], (y0,y1, y2, y3))))
mat = jmatx.stack(jmaty)
det_ideal = ideal(mat.determinant())


incidence0 = F1 + F2 + F1y + F2y + det_ideal
incidence = incidence0 + ideal(p01-XYmin[0],
    p02-XYmin[1],p03-XYmin[2],p12-XYmin[3],p13-XYmin[4],p23-XYmin[5])


incidence = incidence.subs(x0=1)

edge = incidence.elimination_ideal((x0, x1, x2, x3, y0, y1, y2, y3))
edge = edge.change_ring(K[p01, p02, p03, p12, p13, p23])

print "Dimension: ", edge.dimension()-1
print "Degree: ", edge.hilbert_polynomial().coefficients()[1]
#print edge
```

Chow three-fold and secant congruence.

```python
R1.<x0, x1, x2, x3, p01, p02, p03, p12, p13, p23> = K[]

X = matrix(R1,[[x0], [x1], [x2], [x3]])
P = matrix(R1,[[0,p23,-p13,p12],[-p23,0,p03,-p02],[p13,-p03,0,p01],[-p12,p02,-p01,0]])
Pdual = matrix(R1,[[0,-p01,-p02,-p03],[p01,0,-p12,-p13],[p02,p12,0,-p23],[p03,p13,p23,0]])

C1 = C.change_ring(R1)
I = C1 + ideal((P*X).list())
I = I.substitute(x0 = 1)

# Chow
Chow = I.elimination_ideal((x1,x2,x3))
print "Dimension Chow: ", Chow.change_ring(QQ[p01, p02, p03, p12, p13, p23]).dimension()-1
print "Degree Chow: ", Chow.gens()[1].degree()
#print Chow

# Secant
Jmat = jacobian(Chow.gens()[0],(p01, p02, p03, p12, p13, p23)).stack(
jacobian(Chow.gens()[1],(p01, p02, p03, p12, p13, p23)))
JChow = ideal(Jmat.minors(2))
secant = Chow + JChow

point = matrix(R1,[[1],[3],[5],[7]]) # random point and plane
plane = matrix(R1,[[3],[7],[1],[6]])
point_inc = secant + ideal((P*point).list())
plane_inc = secant + ideal((Pdual*plane).list())

print "Order and class of secant: ", (point_inc.change_ring(K[p01, p02, p03, p12, p13, p23]).hilbert_polynomial(), 
                                      plane_inc.change_ring(K[p01, p02, p03, p12, p13, p23]).hilbert_polynomial())
```
